```python
from google.colab import userdata
from langchain_google_genai import ChatGoogleGenerativeAI

# 1. Setup model with your secret key
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    google_api_key=userdata.get('Adi')
)

# 2. Simple list to track context
messages = []

print("Chatbot ready! Type 'exit' to quit.\n")

# 3. Chat loop
while True:
    user_input = input("You: ")
    if user_input.lower() == "exit":
        break

    # Add user message as a tuple: ("user", "text")
    messages.append(("user", user_input))

    # Send conversation history to model
    response = llm.invoke(messages)

    # Add AI response to history as a tuple: ("assistant", "text")
    messages.append(("assistant", response.content))

    # Print response
    print(f"Bot: {response.content}\n")
```
